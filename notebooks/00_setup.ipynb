Colab and Kaggle reset between sessions, so run all cells top to bottom each time.

1. Clone or update the repo
2. Install pinned requirements
3. Point `runs/` at persistent storage (Google Drive on Colab, `/kaggle/working` on Kaggle)
4. Smoke test

```python
import os, sys, subprocess

REPO_URL = "https://github.com/shoimya/Is-the-Judge-the-Bottleneck.git"
IN_COLAB = "google.colab" in sys.modules
IN_KAGGLE = "KAGGLE_KERNEL_RUN_TYPE" in os.environ
WORK = "/content" if IN_COLAB else "/kaggle/working" if IN_KAGGLE else os.getcwd()
REPO = os.path.join(WORK, "Is-the-Judge-the-Bottleneck")
print(sys.version)
print("Colab" if IN_COLAB else "Kaggle" if IN_KAGGLE else "local")
```

```python
# 1. Clone, or pull if already cloned this session
if os.path.isdir(REPO):
    !git -C {REPO} pull --ff-only
else:
    !git clone {REPO_URL} {REPO}
%cd {REPO}
```

```python
# 3. Persistent runs/: logs survive when the session is killed
if IN_COLAB:
    from google.colab import drive
    drive.mount("/content/drive")
    RUNS_TARGET = "/content/drive/MyDrive/Is-the-Judge-the-Bottleneck/runs"
elif IN_KAGGLE:
    RUNS_TARGET = "/kaggle/working/runs"  # saved as notebook output
else:
    RUNS_TARGET = None

if RUNS_TARGET:
    os.makedirs(RUNS_TARGET, exist_ok=True)
    if not os.path.islink("runs"):
        !rm -rf runs
        os.symlink(RUNS_TARGET, "runs")
    print("runs/ ->", os.path.realpath("runs"))
```
